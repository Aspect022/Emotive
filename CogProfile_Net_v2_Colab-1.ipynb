## ⚡ Step 3 — Sub-window + CWT Scalogram Precomputation

```python
SFREQ=128; WIN=128; STEP=32; N_FREQ=64; PAD=32; T_PAD=WIN+2*PAD; W0=5.0
FREQ_MIN=1.0; FREQ_MAX=45.0

def make_subwindows(X, y, pids):
    wins,labs,tids,sids=[],[],[],[]
    for i,(trial,lab,pid) in enumerate(zip(X,y,pids)):
        for s in range(0, trial.shape[1]-WIN+1, STEP):
            wins.append(trial[:,s:s+WIN]); labs.append(lab)
            tids.append(i); sids.append(pid)
    return (np.stack(wins,0).astype(np.float32), np.array(labs,np.int64),
            np.array(tids,np.int64), np.array(sids))

t0=time.time()
X_win,y_win,trial_ids,subj_ids = make_subwindows(X_raw,y,pids)
print(f"Sub-windows: {len(X_win)}  shape={X_win.shape}  elapsed={time.time()-t0:.1f}s")

def build_filters():
    freqs=np.logspace(np.log10(FREQ_MIN),np.log10(FREQ_MAX),N_FREQ)
    filters=[]
    for f in freqs:
        s=W0/(2.0*np.pi*f); M=min(int(6.0*s*SFREQ)|1, T_PAD)
        if M<3: M=3
        t=(np.arange(M)-(M-1)/2.0)/SFREQ; ts=t/s
        psi=(np.pi**-0.25)*np.exp(1j*W0*ts)*np.exp(-0.5*ts**2)/np.sqrt(s)
        if len(psi)<T_PAD:
            pl=(T_PAD-len(psi))//2; pr=T_PAD-len(psi)-pl; psi=np.pad(psi,(pl,pr))
        else:
            st=(len(psi)-T_PAD)//2; psi=psi[st:st+T_PAD]
        filters.append(psi)
    ft=torch.tensor(np.array(filters),dtype=torch.complex64).to(DEVICE)
    return torch.fft.fft(ft,dim=-1)

filters_fft=build_filters()

def compute_all_scalograms(X_win, filters_fft, batch=512):
    N=len(X_win); out=np.empty((N,14,N_FREQ,WIN),dtype=np.float16)
    t0b=time.time()
    for i in range(0,N,batch):
        b=torch.tensor(X_win[i:i+batch],dtype=torch.float32,device=DEVICE)
        padded=F.pad(b,(PAD,PAD),mode="reflect")
        sig_fft=torch.fft.fft(padded,dim=-1)
        prod=sig_fft.unsqueeze(2)*filters_fft.unsqueeze(0).unsqueeze(0).conj()
        conv=torch.fft.ifft(prod,dim=-1)
        power=conv[:,:,:,PAD:PAD+WIN].abs().pow(2)
        power=torch.log1p(power)
        mn=power.amin(-1,keepdim=True); mx=power.amax(-1,keepdim=True)
        power=(power-mn)/torch.clamp(mx-mn,min=1e-8)
        out[i:i+len(b)]=power.to(torch.float16).cpu().numpy()
        if i%(batch*5)==0:
            pct=100*(i+len(b))/N; el=time.time()-t0b
            eta=el/max(i+len(b),1)*(N-i-len(b))
            print(f"  [{pct:5.1f}%] {i+len(b)}/{N}  ETA={eta:.0f}s")
    return out

print("Computing CWT scalograms on GPU...")
SCALOGRAMS=compute_all_scalograms(X_win,filters_fft)
print(f"Done! Shape: {SCALOGRAMS.shape}  total={time.time()-t0:.1f}s")
```

## ✂️ Step 4 — Train/Val/Test Splits

```python
def tier2_split(trial_ids, y_window, seed=42):
    unique_t=np.unique(trial_ids)
    y_t=np.array([y_window[trial_ids==t][0] for t in unique_t])
    sss=StratifiedShuffleSplit(1,test_size=0.1,random_state=seed)
    trv,te=next(sss.split(unique_t,y_t))
    sss2=StratifiedShuffleSplit(1,test_size=0.111,random_state=seed)
    tr,val=next(sss2.split(unique_t[trv],y_t[trv]))
    def wins_of(ts): return np.where(np.isin(trial_ids,unique_t[ts]))[0]
    return {"train_idx":wins_of(trv[tr]), "val_idx":wins_of(trv[val]), "test_idx":wins_of(te)}

def tier3_split(subj_ids, y_window, n_test=5, seed=42):
    gss=GroupShuffleSplit(1,test_size=n_test/len(np.unique(subj_ids)),random_state=seed)
    tr_i,te_i=next(gss.split(np.arange(len(y_window)),y_window,groups=subj_ids))
    np.random.seed(seed)
    val_i=np.random.choice(tr_i,int(0.15*len(tr_i)),replace=False)
    tr_i=np.setdiff1d(tr_i,val_i)
    return {"train_idx":tr_i, "val_idx":val_i, "test_idx":te_i}

SPLITS = {"tier2":tier2_split(trial_ids,y_win), "tier3":tier3_split(subj_ids,y_win)}
sp=SPLITS["tier2"]
print(f"Tier2 > Train: {len(sp['train_idx'])}  Val: {len(sp['val_idx'])}  Test: {len(sp['test_idx'])}")
# ALL indices for unsupervised SimCLR (labels never used in Phase 0)
ALL_IDX = np.arange(len(y_win))
print(f"SimCLR will use ALL {len(ALL_IDX)} windows (unsupervised — no labels needed)")
```

## 🏗️ Step 5 — CogProfile-Net v2 + SimCLR Projection Head

**Architecture additions for self-supervised pre-training:**
- `SimCLRProjectionHead`: 2-layer MLP (128→256→128) attached to encoder output during Phase 0 only
- After pre-training this head is discarded — only the encoder weights are kept
- 6 EEG-specific augmentations used to create contrastive pairs

```python
# ── Branch A: CWT Scalogram CNN ────────────────────────────────
class ScalogramCNN(nn.Module):
    def __init__(self, C=14, F=64, T=128, emb=128, do=0.35):
        super().__init__()
        self.t=nn.Sequential(
            nn.Conv2d(C,32,(1,8),(1,1),(0,4),bias=False),nn.BatchNorm2d(32),nn.ELU())
        self.f=nn.Sequential(
            nn.Conv2d(32,64,(8,1),(1,1),(4,0),groups=32,bias=False),nn.BatchNorm2d(64),
            nn.ELU(),nn.AvgPool2d((2,4)),nn.Dropout(do*0.5))
        self.s=nn.Sequential(
            nn.Conv2d(64,128,(4,4),(1,1),(2,2),bias=False),nn.BatchNorm2d(128),
            nn.ELU(),nn.AvgPool2d((4,4)),nn.Dropout(do))
        with torch.no_grad():
            d=torch.zeros(1,C,F,T); flat=self.s(self.f(self.t(d))).flatten(1).shape[1]
        self.p=nn.Sequential(nn.Flatten(),nn.Linear(flat,emb),nn.LayerNorm(emb))
    def forward(self,x): return self.p(self.s(self.f(self.t(x))))

# ── Branch B: Riemannian SPD Encoder ───────────────────────────
class CovLayer(nn.Module):
    def forward(self,x):
        x=x-x.mean(-1,keepdim=True)
        return torch.bmm(x,x.transpose(1,2))/(x.shape[-1]-1)+torch.eye(x.shape[1],device=x.device)*1e-5

class BiMap(nn.Module):
    def __init__(self,i,o):
        super().__init__(); W=torch.empty(i,o); nn.init.orthogonal_(W); self.W=nn.Parameter(W)
    def forward(self,X): return self.W.T@X@self.W

class ReEig(nn.Module):
    def forward(self,X): L,V=torch.linalg.eigh(X); return V@torch.diag_embed(L.clamp(1e-4))@V.mT

class LogEig(nn.Module):
    def forward(self,X): L,V=torch.linalg.eigh(X); return V@torch.diag_embed(L.clamp(1e-6).log())@V.mT

class RiemannianSPD(nn.Module):
    def __init__(self,C=14,emb=64):
        super().__init__()
        self.cov=CovLayer(); self.b1=BiMap(C,8); self.r1=ReEig()
        self.b2=BiMap(8,6); self.r2=ReEig(); self.log=LogEig()
        self.p=nn.Sequential(nn.Linear(36,emb),nn.LayerNorm(emb))
    def forward(self,x):
        S=self.log(self.r2(self.b2(self.r1(self.b1(self.cov(x))))))
        return self.p(S.flatten(1))

# ── EDL Head ───────────────────────────────────────────────────
class EDLHead(nn.Module):
    def __init__(self,d=128,K=5):
        super().__init__(); self.K=K
        self.fc=nn.Sequential(nn.Linear(d,d//2),nn.ReLU(),nn.Linear(d//2,K))
    def forward(self,z):
        e=F.softplus(self.fc(z)); a=e+1; S=a.sum(1,keepdim=True)
        return {"e":e,"a":a,"prob":a/S,"log_prob":torch.log(a/S+1e-8),"u":self.K/S,"S":S}

def edl_loss(o,y,ep,K=5):
    a=o["a"]; S=o["S"]; yoh=F.one_hot(y,K).float()
    mse=((yoh-a/S).pow(2)+a*(S-a)/(S.pow(2)*(S+1))).sum(1).mean()
    ka=(yoh+(1-yoh)*a)
    lc=torch.lgamma(ka.sum(1))-torch.lgamma(ka).sum(1)
    kl=(lc+((ka-1)*(torch.digamma(ka)-torch.digamma(ka.sum(1,keepdim=True)))).sum(1)).mean()
    return mse+min(1.0,ep/10.0)*kl

# ── DEC ────────────────────────────────────────────────────────
class DEC(nn.Module):
    def __init__(self,K=8,d=128):
        super().__init__(); self.mu=nn.Parameter(torch.randn(K,d))
    @torch.no_grad()
    def init_km(self,embs,seed=42):
        from sklearn.cluster import KMeans
        km=KMeans(n_clusters=self.mu.shape[0],n_init=10,random_state=seed).fit(embs)
        self.mu.data=torch.tensor(km.cluster_centers_,dtype=torch.float32,device=self.mu.device)
        print(f"  DEC centroids initialized (inertia={km.inertia_:.1f})")
    def forward(self,z):
        d2=((z.unsqueeze(1)-self.mu.unsqueeze(0))**2).sum(-1)
        q=(1+d2).pow(-1); return q/q.sum(1,keepdim=True)

def dec_kl(q):
    f=q.sum(0,keepdim=True); p=(q**2/f); p=(p/p.sum(1,keepdim=True)).detach()
    return F.kl_div(q.log(),p,reduction="batchmean")

# ── SimCLR Projection Head (used only during Phase 0) ──────────
class ProjectionHead(nn.Module):
    """
    2-layer MLP: 128 → 256 → 128 with BN.
    Attached during SimCLR pre-training, discarded before supervised training.
    """
    def __init__(self, in_dim=128, hidden_dim=256, out_dim=128):
        super().__init__()
        self.net=nn.Sequential(
            nn.Linear(in_dim,hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim,out_dim)
        )
    def forward(self,x): return self.net(x)

def nt_xent_loss(z1, z2, temperature=0.5):
    """
    NT-Xent (Normalized Temperature-scaled Cross Entropy) — the SimCLR loss.
    z1, z2: two views of the same batch, shape (N, D).
    For each sample i, its positive pair is sample i in the other view.
    All other 2N-2 samples in the batch are negatives.
    """
    z1 = F.normalize(z1, dim=1)
    z2 = F.normalize(z2, dim=1)
    N  = z1.shape[0]
    z  = torch.cat([z1, z2], dim=0)          # (2N, D)
    sim = torch.mm(z, z.T) / temperature      # (2N, 2N) cosine sim matrix
    # Mask out self-similarity diagonal
    mask = torch.eye(2*N, dtype=torch.bool, device=z.device)
    sim.masked_fill_(mask, float('-inf'))
    # Positive pairs: (i, i+N) and (i+N, i)
    labels = torch.cat([torch.arange(N,2*N), torch.arange(N)]).to(z.device)
    return F.cross_entropy(sim, labels)

# ── Full CogProfile-Net v2 ─────────────────────────────────────
class CogProfileNet(nn.Module):
    def __init__(self,K=5,emb=128,n_clust=8,do=0.35):
        super().__init__(); self.K=K
        self.a=ScalogramCNN(emb=emb,do=do)
        self.b=RiemannianSPD(emb=emb//2)
        self.pa=nn.Linear(emb,emb); self.pb=nn.Linear(emb//2,emb)
        self.gate=nn.Sequential(nn.Linear(emb*2,emb),nn.Sigmoid())
        self.ln=nn.LayerNorm(emb)
        self.bmlp=nn.Sequential(nn.Linear(3,16),nn.ReLU(),nn.Linear(16,emb),nn.Sigmoid())
        self.aux=nn.Linear(emb,K)
        self.dec=DEC(n_clust,emb)
        self.edl=EDLHead(emb,K)

    def encode(self,sc,rw,bh):
        """Shared encoder — used by both SimCLR and supervised training."""
        za=self.a(sc); zb=self.b(rw)
        a_=self.pa(za); b_=self.pb(zb)
        g=self.gate(torch.cat([a_,b_],1))
        z=self.ln(g*a_+(1-g)*b_)
        z=z+self.bmlp(bh)*z
        return z, za, zb

    def forward(self,sc,rw,bh):
        z,za,zb=self.encode(sc,rw,bh)
        return {"z":z,"za":za,"zb":zb,"q":self.dec(z),"edl":self.edl(z),"aux":self.aux(z)}

    def compute_loss(self,o,y,ep,ld=0.01,lo=0.05,la=0.2):
        el=edl_loss(o["edl"],y,ep)
        dl=dec_kl(o["q"])
        d=min(o["za"].shape[1],o["zb"].shape[1])
        ol=(F.normalize(o["za"][:,:d],1)*F.normalize(o["zb"][:,:d],1)).sum(1).pow(2).mean()
        al=F.cross_entropy(o["aux"],y,label_smoothing=0.1)
        return el+ld*dl+lo*ol+la*al

model=CogProfileNet().to(DEVICE)
proj_head=ProjectionHead().to(DEVICE)   # SimCLR projection head
n_p=sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"CogProfile-Net v2: {n_p:,} params | ProjectionHead: {sum(p.numel() for p in proj_head.parameters()):,} params")
# Quick sanity check
with torch.no_grad():
    sc=torch.randn(4,14,64,128).to(DEVICE); rw=torch.randn(4,14,128).to(DEVICE)
    bh=torch.zeros(4,3).to(DEVICE)
    z,_,_=model.encode(sc,rw,bh)
    proj=proj_head(z)
    print(f"Encoder z: {z.shape}  Projected: {proj.shape}")
print("✅ Architecture OK")
```

## 📦 Step 6 — Datasets & DataLoaders

```python
# ── EEG Augmentation functions ─────────────────────────────────
def augment_pair(sc, rw):
    """
    Apply 2 independent random augmentations to produce a contrastive pair.
    6 augmentation types, each applied independently for view 1 and view 2.
    Returns: (sc1, rw1), (sc2, rw2)
    """
    def aug_single(sc, rw):
        sc=sc.copy(); rw=rw.copy()
        # 1. Temporal flip (time reversal)
        if np.random.rand()<0.5:
            rw=rw[:,::-1].copy(); sc=sc[:,:,::-1].copy()
        # 2. Gaussian noise
        if np.random.rand()<0.5:
            rw+=np.random.normal(0,0.01,rw.shape).astype(np.float32)
            sc=np.clip(sc+np.random.normal(0,0.01,sc.shape).astype(np.float32),0,1)
        # 3. Amplitude scaling
        if np.random.rand()<0.5:
            scale=np.random.uniform(0.8,1.2)
            rw=rw*scale
            sc=np.clip(sc*scale,0,1)
        # 4. Channel dropout (zero 1 random channel)
        if np.random.rand()<0.3:
            ch=np.random.randint(0,rw.shape[0])
            rw[ch]=0.0; sc[ch]=0.0
        # 5. Temporal shift (circular shift up to 16 samples)
        if np.random.rand()<0.4:
            shift=np.random.randint(-16,17)
            rw=np.roll(rw,shift,axis=1)
            sc=np.roll(sc,shift,axis=2)
        # 6. Frequency band masking (zero out a random freq band in scalogram)
        if np.random.rand()<0.3:
            f_start=np.random.randint(0,48); f_end=f_start+np.random.randint(4,16)
            sc[:,f_start:min(f_end,64),:]=0.0
        return sc, rw
    sc1,rw1=aug_single(sc,rw)
    sc2,rw2=aug_single(sc,rw)
    return (sc1,rw1),(sc2,rw2)

# ── SimCLR Dataset (unlabeled, all windows) ─────────────────────
class SimCLRDataset(Dataset):
    """Returns two augmented views of each window. No labels used."""
    def __init__(self,S,X,idx):
        self.S=S; self.X=X; self.idx=idx
        self.beh=np.array([0.5,0.5,0.5],dtype=np.float32)
    def __len__(self): return len(self.idx)
    def __getitem__(self,i):
        r=self.idx[i]
        sc=np.array(self.S[r],dtype=np.float32)
        rw=np.array(self.X[r],dtype=np.float32)
        (sc1,rw1),(sc2,rw2)=augment_pair(sc,rw)
        bh=torch.from_numpy(self.beh)
        return (torch.from_numpy(sc1),torch.from_numpy(rw1),bh,
                torch.from_numpy(sc2),torch.from_numpy(rw2),bh)

# ── Supervised Dataset ─────────────────────────────────────────
class EEGDataset(Dataset):
    def __init__(self,S,X,y,idx,aug=False):
        self.S=S; self.X=X; self.y=y; self.idx=idx; self.aug=aug
        self.beh=np.array([0.5,0.5,0.5],dtype=np.float32)
    def __len__(self): return len(self.idx)
    def __getitem__(self,i):
        r=self.idx[i]
        sc=np.array(self.S[r],dtype=np.float32)
        rw=np.array(self.X[r],dtype=np.float32)
        if self.aug:
            (sc,rw),_=augment_pair(sc,rw)  # reuse augment, take only view 1
        return (torch.from_numpy(sc),torch.from_numpy(rw),
                torch.from_numpy(self.beh),torch.tensor(self.y[r],dtype=torch.long))

BATCH=256; sp=SPLITS["tier2"]
# SimCLR loader uses ALL windows
simclr_loader=DataLoader(SimCLRDataset(SCALOGRAMS,X_win,ALL_IDX),
                         batch_size=256,shuffle=True,num_workers=2,pin_memory=True,drop_last=True)
# Supervised loaders
train_loader=DataLoader(EEGDataset(SCALOGRAMS,X_win,y_win,sp["train_idx"],True),
                        BATCH,True,num_workers=2,pin_memory=True)
val_loader  =DataLoader(EEGDataset(SCALOGRAMS,X_win,y_win,sp["val_idx"]),
                        BATCH,False,num_workers=2,pin_memory=True)
test_loader =DataLoader(EEGDataset(SCALOGRAMS,X_win,y_win,sp["test_idx"]),
                        BATCH,False,num_workers=2,pin_memory=True)
print(f"SimCLR pool: {len(simclr_loader.dataset)} windows")
print(f"Supervised — Train: {len(train_loader.dataset)}  Val: {len(val_loader.dataset)}  Test: {len(test_loader.dataset)}")
```

## 🏋️ Step 7 — Four-Phase Training

| Phase | Epochs | What happens |
|---|---|---|
| **Phase 0** | 30 | SimCLR self-supervised pre-training on ALL windows (no labels) |
| **Phase 1** | 30 | Supervised encoder fine-tuning (no DEC) |
| **Phase 2a** | — | K-Means initialise DEC centroids |
| **Phase 2b** | 80 | Joint DEC + EDL training |


```python
@torch.no_grad()
def evaluate(model,loader):
    model.eval(); ps,ls,us=[],[],[]
    for sc,rw,bh,lb in loader:
        sc,rw,bh=sc.to(DEVICE),rw.to(DEVICE),bh.to(DEVICE)
        o=model(sc,rw,bh)
        ps.extend(o["edl"]["prob"].argmax(1).cpu().numpy())
        ls.extend(lb.numpy())
        us.extend(o["edl"]["u"].squeeze(1).cpu().numpy())
    ps,ls=np.array(ps),np.array(ls)
    acc=(ps==ls).mean()
    f1s=[]
    for k in range(5):
        tp=((ps==k)&(ls==k)).sum(); fp=((ps==k)&(ls!=k)).sum(); fn=((ps!=k)&(ls==k)).sum()
        p=tp/(tp+fp+1e-8); r=tp/(tp+fn+1e-8); f1s.append(2*p*r/(p+r+1e-8))
    return float(acc), float(np.mean(f1s)), float(np.array(us).mean())

CKPT="/content/cogprofile_best.pt"
best_acc=0.0; hist=[]; T0=time.time(); pat=0

# ═══════════════════════════════════════════════════════════════
# PHASE 0: SimCLR Self-Supervised Pre-training
# Goal: learn generalised EEG representations without any labels
# Uses ALL windows — maximises the self-supervised signal
# ═══════════════════════════════════════════════════════════════
SIMCLR_EPOCHS=30
print("="*60)
print(f"PHASE 0: SimCLR self-supervised pre-training ({SIMCLR_EPOCHS} epochs)")
print(f"  Using {len(simclr_loader.dataset)} windows | Temperature=0.5")
print("  [No labels used — purely unsupervised]")

# Separate optimiser for pre-training (encoder + projection head)
simclr_params=list(model.parameters())+list(proj_head.parameters())
simclr_opt=torch.optim.AdamW(simclr_params,lr=1e-3,weight_decay=1e-4)
simclr_sched=torch.optim.lr_scheduler.CosineAnnealingLR(simclr_opt,SIMCLR_EPOCHS,eta_min=1e-5)

for ep in range(1,SIMCLR_EPOCHS+1):
    model.train(); proj_head.train()
    tloss=0.0; nb=0
    for sc1,rw1,bh1,sc2,rw2,bh2 in simclr_loader:
        sc1,rw1,bh1=sc1.to(DEVICE),rw1.to(DEVICE),bh1.to(DEVICE)
        sc2,rw2,bh2=sc2.to(DEVICE),rw2.to(DEVICE),bh2.to(DEVICE)
        simclr_opt.zero_grad()
        # Encode both views
        z1,_,_=model.encode(sc1,rw1,bh1)
        z2,_,_=model.encode(sc2,rw2,bh2)
        # Project to contrastive space
        p1=proj_head(z1); p2=proj_head(z2)
        # NT-Xent loss
        loss=nt_xent_loss(p1,p2,temperature=0.5)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(simclr_params,1.0)
        simclr_opt.step()
        tloss+=loss.item(); nb+=1
    simclr_sched.step()
    if ep%5==0 or ep==1:
        el=time.time()-T0
        print(f"  SimCLR Ep {ep:3d} | NT-Xent loss={tloss/nb:.4f} | {el:.0f}s")

# Discard projection head — encoder weights are now pre-trained
del proj_head, simclr_opt, simclr_sched
torch.cuda.empty_cache()
print(f"\n✅ SimCLR pre-training complete in {(time.time()-T0)/60:.1f} min")
print("   Projection head discarded — encoder weights retained")

# ═══════════════════════════════════════════════════════════════
# PHASE 1: Supervised encoder fine-tuning (no DEC)
# ═══════════════════════════════════════════════════════════════
PHASE1_EPOCHS=30; PHASE2_EPOCHS=80; WARMUP=5
opt=torch.optim.AdamW(model.parameters(),lr=5e-4,weight_decay=1e-4)  # lower LR — encoder already pre-trained
warmup_sched=torch.optim.lr_scheduler.LinearLR(opt,start_factor=0.2,end_factor=1.0,total_iters=WARMUP)
cosine_sched=torch.optim.lr_scheduler.CosineAnnealingLR(opt,PHASE1_EPOCHS+PHASE2_EPOCHS-WARMUP,eta_min=1e-5)
sched=torch.optim.lr_scheduler.SequentialLR(opt,schedulers=[warmup_sched,cosine_sched],milestones=[WARMUP])

print()
print("="*60)
print(f"PHASE 1: Supervised fine-tuning (no DEC, {PHASE1_EPOCHS} epochs)")
print("  [Lower LR=5e-4 since encoder already has good representations]")
for ep in range(1,PHASE1_EPOCHS+1):
    model.train()
    for sc,rw,bh,lb in train_loader:
        sc,rw,bh,lb=sc.to(DEVICE),rw.to(DEVICE),bh.to(DEVICE),lb.to(DEVICE)
        opt.zero_grad()
        o=model(sc,rw,bh)
        ce =F.nll_loss(o["edl"]["log_prob"],lb,label_smoothing=0.1)
        aux=F.cross_entropy(o["aux"],lb,label_smoothing=0.1)
        (ce+0.2*aux).backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(),1.0); opt.step()
    sched.step()
    if ep%5==0:
        va,vf,vu=evaluate(model,val_loader)
        lr_now=opt.param_groups[0]['lr']
        print(f"  Ep {ep:3d} | val_acc={va:.4f} | macro_f1={vf:.4f} | lr={lr_now:.2e}")
        if va>best_acc:
            best_acc=va; torch.save(model.state_dict(),CKPT)
            print(f"    ✓ New best saved ({best_acc:.4f})")

# ═══════════════════════════════════════════════════════════════
# PHASE 2a: Init DEC centroids via K-Means
# ═══════════════════════════════════════════════════════════════
print()
print("PHASE 2a: Initialising DEC centroids via K-Means...")
model.eval(); embs=[]
with torch.no_grad():
    for sc,rw,bh,_ in train_loader:
        sc,rw,bh=sc.to(DEVICE),rw.to(DEVICE),bh.to(DEVICE)
        embs.append(model(sc,rw,bh)["z"].cpu().numpy())
model.dec.init_km(np.concatenate(embs))

# ═══════════════════════════════════════════════════════════════
# PHASE 2b: Joint DEC + EDL training
# ═══════════════════════════════════════════════════════════════
print()
print(f"PHASE 2b: Joint DEC + EDL training ({PHASE2_EPOCHS} epochs)")
for ep in range(1,PHASE2_EPOCHS+1):
    model.train(); tloss=0.0; nb=0
    for sc,rw,bh,lb in train_loader:
        sc,rw,bh,lb=sc.to(DEVICE),rw.to(DEVICE),bh.to(DEVICE),lb.to(DEVICE)
        opt.zero_grad()
        o=model(sc,rw,bh)
        loss=model.compute_loss(o,lb,ep)
        loss.backward(); torch.nn.utils.clip_grad_norm_(model.parameters(),1.0); opt.step()
        tloss+=loss.item(); nb+=1
    sched.step()
    if ep%5==0 or ep==1:
        va,vf,vu=evaluate(model,val_loader)
        el=time.time()-T0
        print(f"  Ep {ep:3d} | loss={tloss/nb:.4f} | val_acc={va:.4f} | f1={vf:.4f} | u={vu:.3f} | {el:.0f}s")
        hist.append({"epoch":ep,"acc":va,"f1":vf})
        if va>best_acc:
            best_acc=va; pat=0; torch.save(model.state_dict(),CKPT)
            print(f"    ✓ New best saved ({best_acc:.4f})")
        else:
            pat+=1
            if pat>=5: print(f"  Early stopping at epoch {ep}"); break

print()
print(f"Training complete in {(time.time()-T0)/60:.1f} min")
print(f"Best val accuracy: {best_acc:.4f}")
```

## 📊 Step 8 — Final Evaluation

```python
model.load_state_dict(torch.load(CKPT,map_location=DEVICE,weights_only=True))
NAMES=["Mental Arithmetic","Pattern Recognition","Working Memory","Reading Comprehension","Sustained Attention"]

print("="*60)
print("FINAL RESULTS ACROSS ALL TIERS")
print("="*60)
for tier,sp_t in SPLITS.items():
    ds=EEGDataset(SCALOGRAMS,X_win,y_win,sp_t["test_idx"])
    lo=DataLoader(ds,256,num_workers=2,pin_memory=True)
    a,f,u=evaluate(model,lo)
    print(f"  [{tier.upper():6s}] Accuracy={a:.4f}  Macro-F1={f:.4f}  Uncertainty={u:.3f}")

all_p,all_l=[],[]
with torch.no_grad():
    for sc,rw,bh,lb in test_loader:
        sc,rw,bh=sc.to(DEVICE),rw.to(DEVICE),bh.to(DEVICE)
        o=model(sc,rw,bh)
        all_p.extend(o["edl"]["prob"].argmax(1).cpu().numpy())
        all_l.extend(lb.numpy())

print()
print("Tier-2 Classification Report:")
print(classification_report(np.array(all_l),np.array(all_p),target_names=NAMES))
```
